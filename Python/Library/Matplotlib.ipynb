# Piechart

```python
y = np.array([35, 25, 25, 15])
mylabels = ["Bery", "Bananas", "Cherries", "Apples"]

plt.pie(y,labels=mylabels,startangle=90)  # start from angle of 90 (default )
plt.show() 
```

```python
y = np.array([35, 25, 25, 15])
mylabels = ["Bery", "Bananas", "Cherries", "Apples"]
myexplode = [0, 0, 0.5, 0]
mycolors = ["m", "hotpink", "b", "#4CAF50"]

plt.pie(y,labels=mylabels,startangle=90,explode=myexplode,shadow=True,colors=mycolors)  # start from angle of 90 (default 0)
plt.legend(title = "Four Fruits:")
plt.show() 
```

```python
from sklearn.impute import SimpleImputer
```

```python
df=pd.read_csv("country_wise_latest.csv")
df.head(10)
```

```python
df.drop(['Deaths / 100 Cases','Recovered / 100 Cases','Deaths / 100 Recovered'],axis=1,inplace=True)
df.rename(columns={'Deaths':"TapakGye"})
```

```python
# Impute missing values with a constant value (default: 0 for numbers, empty string for strings)
imputer=SimpleImputer(strategy='constant')
df2=pd.DataFrame(imputer.fit_transform(df),columns=df.columns)
df2
```

```python
df3=df2.groupby(['WHO Region','Country/Region'])[['Confirmed','Deaths','Recovered','Active']].sum()
df3=df3.reset_index()
df3
```

```python
WHO_Region=df3['WHO Region'].unique()
print(len(WHO_Region))
```

```python
Country_Region=df3['Country/Region'].unique()
print(len(Country_Region))
```

```python
df3.describe()
```

```python
for idx in range(0,len(WHO_Region)):
    C=df3[df3['WHO Region']==WHO_Region[idx]].reset_index()
    # print(C)
    plt.figure(figsize=(12, 6))
    # In plt.scatter first parameter is x-axis variables and second is y-axis variables
    plt.scatter(C["Country/Region"],C['Confirmed'],color='blue',label='Confirmed')
    plt.scatter(C["Country/Region"],C['Deaths'],color='red',label='Deaths')
    plt.scatter(C["Country/Region"],C['Recovered'],color='green',label='Recovered')
    plt.title(WHO_Region[idx])
    plt.xlabel("Country")
    plt.ylabel("Number of cases")
    plt.xticks(rotation=90)  # Rotate country names for better readability
    plt.legend()
    plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.scatter(df3["Country/Region"],df3['Confirmed'],color='blue',label='Confirmed')
plt.scatter(df3["Country/Region"],df3['Deaths'],color='red',label='Deaths')
plt.scatter(df3["Country/Region"],df3['Recovered'],color='green',label='Recovered')
plt.title("World")
plt.xlabel("Country")
plt.ylabel("No. of Cases")
plt.xticks(rotation=90)
plt.legend()
plt.show()
```
