# numpy slicing 

## A[start:end:step]

```python
A=np.arange(10)
```

```python
A[3:9]
```

```python
B=A[::-1]    # this only makes refrence
B
```

```python
B[0]=10
B
```

```python
A       ## See above 2 cell , i.e B is sliceing of A but i f we change in B it reflect to A , this is diifrence between in List and array
```

```python
B=A[3:8].copy()
B
```

```python
A=np.arange(100)
A[3]=1200
A
```

```python
idx = np.argwhere(A==1200)[0][0]
A[idx]=3
A
```
