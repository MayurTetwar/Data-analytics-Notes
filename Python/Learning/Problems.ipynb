```python
def getDataFromUser():
    D = {}
    while True:
        studentId = input("Enter students ID")
        if studentId in D:
            print(f'{studentId} is already inserted')
            continue;
        marksList = input("Enter the marks by comma sepreted values: ")
        moreStudents = input("Enter no for quit insertion: ")    
        D[studentId]=marksList.split(",")
        if moreStudents.lower() == "no":
            return D;

```

```python
studentsData=getDataFromUser()
```

```python
studentsData
```

```python
def getAvgMarks(D):
    for id in D:
        total=0
        numbers=0
        for mark in D[id]:
            total+=int(mark)
            numbers+=1
        print(f'Averge Marks of id {id} is {(total/numbers)}')
```

```python
getAvgMarks(studentsData)
```
