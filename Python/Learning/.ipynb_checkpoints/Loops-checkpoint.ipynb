```python
import time
## After such ampunt of try system get block for perticular of time
wait_time=1
maxtry=5
attempts=0

while(attempts<maxtry):
    print(f'{attempts+1} , {wait_time}')
    time.sleep(wait_time)
    wait_time*=2
    attempts+=1
```

```python
list=[68,3,6,8,9,3,5,6,-2,0,4]
i=0
j=0
for i in range(0,len(list)):
    num=list[i]
    ind=i
    for j in range(i+1,len(list)):
        if(list[j]<num):
            num=list[j]
            ind=j
    temp=list[i]
    list[i]=list[ind]
    list[ind]=temp
list
```
