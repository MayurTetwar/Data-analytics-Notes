```python
def isPoli(str):
    i=0;
    j=len(str)-1
    while(i<j):
        if(str[i]!=str[j]):
            return False
        i+=1
        j-=1
    return True

print(isPoli("MayaM"))
print(isPoli("Mayur"))
```

```python
def fibo(num):
    if(num==1):
        return 0
    elif(num==2):
        return 1
    return fibo(num-1)+fibo(num-2)

for i in range(1,10):
    print(fibo(i))
```

```python

dic={
    '-':1,
    '+':2,
    '*':3,
    '/':4
}

def calculate(str):
    stk=[]
    ans=[]
    ind=0;
    for ch in str:
        if(ch>='0' and ch<='9'):
            ans.append(int(ch))
        else:
            last=len(stk)-1
            while(last>=0 and dic[stk[last]]>dic[ch]):
                num2=ans.pop()
                num1=ans.pop()
                if(stk[last]=='+'):  
                    ans.append(num1+num2)
                elif(stk[last]=='-'):
                    ans.append(num1-num2)
                elif(stk[last]=='*'):  
                    ans.append(num1*num2)
                elif(stk[last]=='/'):  
                     ans.append(num1/num2)
                stk.pop()
                last-=1
            stk.append(ch)

    last=len(stk)-1
    while(len(stk)>0):
        num2=ans.pop()
        num1=ans.pop()
        if(stk[last]=='+'):  
            ans.append(num1+num2)
        elif(stk[last]=='-'):
            ans.append(num1-num2)
        elif(stk[last]=='*'):  
            ans.append(num1*num2)
        elif(stk[last]=='/'):  
            ans.append(num1/num2)
        stk.pop()
        last-=1
    print(ans)
    
calculate("3*2+3/2*4+5")
```

```python
def buble_sort(arr):
    for n in range(len(arr) - 1, 0, -1):
        swapped = False  
        for i in range(n):
            if arr[i] > arr[i + 1]:
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
                swapped = True
        if not swapped:
            break
def insert_sort(arr):
    n = len(arr)
    if n <= 1:
        return
    for i in range(1, n):
        key = arr[i]
        j = i-1
        while j >= 0 and key < arr[j]:
            arr[j+1] = arr[j]  
            j -= 1
        arr[j+1] = key
def partition(array, low, high):
    pivot = array[high]
    i = low - 1
    for j in range(low, high):
        if array[j] <= pivot:
            i = i + 1
            (array[i], array[j]) = (array[j], array[i])
    (array[i + 1], array[high]) = (array[high], array[i + 1])
    return i + 1

def quickSort(array, low, high):
    if low < high:
        pi = partition(array, low, high)
        quickSort(array, low, pi - 1)
        quickSort(array, pi + 1, high)

from numpy import random
x=random.randint(100, size=(6))
print(x)
choice=int(input("Enter the choice: 1.buble sort 2.Quick sort 3.Insertion sort"))
if choice==1 :
    buble_sort(x)
elif choice==2:
    insert_sort(x)
elif choice ==3 :
   quickSort(x, 0, size - 1)
else :
    print("Invlid input")
print(f"sorted array:{x}")
```

```python
class BankAcc:
    def __init__(self,AccNo=0,Mon=0):
        self.AccNo=AccNo
        self.Mon=Mon
    def deposit(self): 
        income=int(input("Enter Amount to Deposit:"))
        if income > 0:
            self.Mon += income
        else:
            print("Deposit amount must be positive.")
    def withdrawal(self):
        income=int(input(f"You have {self.Mon} money \n Enter Withdrawal Amount:"))
        if(self.Mon<income or income<=0):
            print("Invlid Amount Entered!...")
        else:
            self.Mon-=income
    def checking(self):
        print(f'You Have {self.Mon} Balanced!...')
        
def main():
    AccNo=int(input("Enter Bank Account Number:"))
    Mon=int(input("Enter Money In your Acount:"))
    Obj=BankAcc(AccNo,Mon)
    while(True):
        option=int(input("Enter 1 for Deposit \n Enter 2 for WithDrawal \n 3 for Cheaking Account balanced \n 4 for Quit :"))
        if(option==1):
            Obj.deposit()
        elif(option==2):
            Obj.withdrawal()
        elif(option==3):
            Obj.checking()
        else:
            print("ENd.....")
            return;
if(__name__=="__main__"):
    main()
```

```python
class Node:
    def __init__(self,task):
        self.task=task
        self.next=None

class ToDoList:
    def __init__(self):
        self.head=None

    def AddTask(self,task):
        new_Node=Node(task)
        if(self.head is None):
            self.head=new_Node
        else:
            curr=self.head
            while(curr.next is not None):
                curr=curr.next
            curr.next=new_Node
        print(f'Task {task} is Add!...')

    def RemoveTask(self,pos):
        if(self.head is None):
            print("Todolist is empty!...")
            return
        if pos==1:
            task=self.head.task
            self.head=self.head.next
            print(f'Task {task} is Remove!...')
            return

        curr=self.head
        count=0
        while curr and count<pos-2:
            curr=curr.next
            pos+=1

        if(curr is None):
            print("Invalid position!....")
            return
        else:
            task=curr.next.task
            curr.next=curr.next.next
            print(f'Task {task} is Remove!....')

    def Display(self):
        if self.head is None:
            print(f'TodoList is Empty!....')
            return

        curr=self.head
        pos=1
        while curr:
            print(f'{pos}. {curr.task}')
            curr=curr.next
            pos+=1


todo = ToDoList()
todo.AddTask("Buy groceries")
todo.AddTask("Finish Code")
todo.AddTask("Clean the house")

todo.Display()
todo.RemoveTask(2)
todo.Display()
```

```python
graph={}

def addFriend(friend):
    if friend not in graph:
        graph[friend]=set()
        
def addFriendShip(fri1,fri2):
    if fri1 not in graph:
        addFriend(fri1)
    if fri2 not in graph:
        addFriend(fri2)
    graph[fri1].add(fri2)
    graph[fri2].add(fri1)

def displayFriend(fri):
    if fri not in graph:
        print(f"{fri} is not in Network!...\n")
        return
    print(f'{fri}s frined are ')
    for friend in graph[fri]:
        print(f'{friend} ',end=' ')
    print(f'\n')

def displaymutualFri(fri1,fri2):
    if fri1 in graph and fri2 in graph:
        print(f'Mutual frineds of {fri1} and {fri2} are',end=" " )
        set=graph[fri1].intersection(graph[fri2])
        if len(set)==0:
            print("None")
            return
        for friend in set:
            print(f'{friend} ',end=' ')
        print
        return
    print(f'from {fri1} or {fri2} one of the [erson is not present in network!....\n')

addFriendShip("Mayur","Tejas")
addFriendShip("Tejas","Bhavesh")
addFriendShip("Bhavesh","Mayur")
addFriendShip("Mayur","Harsh")
addFriendShip("Harsh","Akshat")

displayFriend("Mayur")
displayFriend("Akshat")

displaymutualFri("Mayur","Harsh")
displaymutualFri("Tejas","Bhavesh")
```
